```sql
CREATE OR REPLACE VIEW workspace.default.vw_daily_funnel AS
SELECT
  event_date,
  total_events,
  unique_users,
  sessions,
  views,
  carts,
  remove_from_cart,
  purchases,
  -- simple conversion rates
  ROUND(purchases / NULLIF(views, 0), 4)      AS view_to_purchase_rate,
  ROUND(purchases / NULLIF(sessions, 0), 4)   AS session_to_purchase_rate
FROM workspace.default.gold_funnel_metrics;
```
